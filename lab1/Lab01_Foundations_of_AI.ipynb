```python
import os
import cv2
import torch
import ultralytics
import matplotlib
import numpy as np

# 1. Ensure Directory Structure
directories = [
    "data/raw", "data/processed", "models",
    "notebooks", "src", "configs", "logs"
]
for folder in directories:
    os.makedirs(folder, exist_ok=True)

# 2. Write .gitignore
gitignore_content = """# Python
__pycache__/
*.py[cod]

# Jupyter Notebook
.ipynb_checkpoints

# Virtual Environments
.venv/
venv/
aipdd_env/

# Binaries & Model Weights
*.pt
*.onnx
*.h5
"""
with open(".gitignore", "w") as f:
    f.write(gitignore_content)

# 3. Make .gitignore visible in Colab file tree (copies it as gitignore.txt for easy download)
with open("gitignore.txt", "w") as f:
    f.write(gitignore_content)

# 4. Write requirements.txt
requirements_content = f"""numpy=={np.__version__}
opencv-python=={cv2.__version__}
matplotlib=={matplotlib.__version__}
ultralytics=={ultralytics.__version__}
torch=={torch.__version__}
"""
with open("requirements.txt", "w") as f:
    f.write(requirements_content)

print("Files successfully generated!")
```
